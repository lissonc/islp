## The workflow for every lab

1. **Click the lab's "Open in Colab" link** (in the repository README, on the
   learning platform, or the badge at the top of each notebook). The clean,
   read-only master copy opens in Colab. Nothing you do can break it.
2. **Save your own copy first:** in Colab choose **File → Save a copy in
   Drive**. Your copy lands in the folder *Colab Notebooks* in your Google
   Drive — that copy is where all your work lives. If you skip this step you
   can still run cells, but your changes are **not saved**.
3. **Run the setup cell at the top — first thing, every session.** Each lab
   starts with a cell labelled *Google Colab setup*. It installs the course
   software into the fresh Colab machine and takes about 1–2 minutes.
4. **Work top-to-bottom.** Cells build on earlier cells. If results look
   strange, use **Runtime → Run all** (or *Run before*) to re-run cells in
   order.
5. **Continuing later:** open *your* copy from Google Drive or from
   [colab.research.google.com](https://colab.research.google.com) → *Recent*.
   Your text and saved outputs are still there, but the machine behind the
   notebook is new — so run the setup cell (and the cells above where you
   stopped) again before continuing.
6. **Made a mess of your copy?** No problem: click the original lab link
   again and you get a pristine copy. Save it to Drive under a new name and
   carry on. You can always get back to a clean start.

## Limits of the free Colab tier

Colab's free tier is fully sufficient for this course, but know its limits:

- **Sessions are temporary.** A session disconnects after roughly 90 minutes
  of inactivity and after at most about 12 hours. Everything *installed or
  downloaded* on the machine disappears then — that is why the setup cell
  must be re-run each session. Your notebook itself (text and outputs) is
  safe in your Drive copy.
- **Usage limits are dynamic.** Google adjusts free-tier limits based on
  demand; heavy use over several days can lead to temporary cool-downs.
  Normal weekly lab work stays well below these limits — but **start
  assignments early** rather than in one marathon session before the
  deadline.
- **GPUs are not guaranteed.** Only the Chapter 10 deep-learning lab benefits
  from a GPU (**Runtime → Change runtime type → T4 GPU**). On the free tier a
  GPU is usually, but not always, available — another reason not to leave
  that lab to the last minute.
- **Drive storage counts against your quota** (15 GB on a free account).
  Notebooks are tiny; this will not be a problem in practice.

A paid Colab plan is **not** required for this course.
